Time-Binning to create Historical Data, here we bin the time series data into 15-minute intervals and count the number of pickups in each interval.
The number of pickups can fluctuate significantly and we can predict with high accuracy for next time interval. Hence, we will be using Smoothing for more accurate predictions. We will be using Moving Average (MA) and Exponential Weightage Moving Average (EWMA) for smoothing the data.


In Moving Average, window size means the number of data points to consider for calculating the average. Higher window size means more weightage to the older data points thus smoother curves and lower window size means more weightage to the recent data points thus sharper curves.


In Exponential Weightage Moving Average, high alpha means more weightage to the recent data points and low alpha means more weightage to the older data points.

In conclusion, we will calculate number of pickups for each region and for each 15 minute time interval. then smooth the data.

# Importing Libraries & Dataset

```python
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import MiniBatchKMeans
import matplotlib.pyplot as plt
import dask.dataframe as dd
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# paths for the three dfs
df_jan_path = "../data/raw/yellow_tripdata_2016-01.csv"
df_feb_path = "../data/raw/yellow_tripdata_2016-02.csv"
df_mar_path = "../data/raw/yellow_tripdata_2016-03.csv"

# load the dataframes
columns_to_use = ['trip_distance', 'tpep_pickup_datetime', 'pickup_longitude', 'pickup_latitude','dropoff_longitude', 'dropoff_latitude', 'fare_amount']
df_jan = dd.read_csv(df_jan_path, assume_missing=True, usecols=columns_to_use, parse_dates=["tpep_pickup_datetime"])
df_feb = dd.read_csv(df_feb_path, assume_missing=True, usecols=columns_to_use, parse_dates=["tpep_pickup_datetime"])
df_mar = dd.read_csv(df_mar_path, assume_missing=True, usecols=columns_to_use, parse_dates=["tpep_pickup_datetime"])
# parse_dates to set the tpep_pickup_datetime column as datetime type
```

```python
df_jan
```

```python
# concat the three dataframes as one

df_final = dd.concat([df_jan, df_feb, df_mar], axis=0)
```

# Working

```python
# set the values of coordinates

min_latitude = 40.60
max_latitude = 40.85

min_longitude = -74.05
max_longitude = -73.70

min_fare_amount_val = 0.50
max_fare_amount_val = 81.0

min_trip_distance_val = 0.25
max_trip_distance_val = 24.43
```

```python
# select data points within the given ranges

df_final = df_final.loc[(df_final["pickup_latitude"].between(min_latitude, max_latitude, inclusive="both")) & 
(df_final["pickup_longitude"].between(min_longitude, max_longitude, inclusive="both")) & 
(df_final["dropoff_latitude"].between(min_latitude, max_latitude, inclusive="both")) & 
(df_final["dropoff_longitude"].between(min_longitude, max_longitude, inclusive="both")), :]
```

```python
# select data points within the given ranges

df_final = df_final.loc[(df_final["fare_amount"].between(min_fare_amount_val, max_fare_amount_val, inclusive="both")) & 
(df_final["trip_distance"].between(min_trip_distance_val, max_trip_distance_val, inclusive="both"))]
```

```python
# drop columns from the data

df_final = df_final.drop(columns=['trip_distance', 'dropoff_longitude', 'dropoff_latitude', 'fare_amount'])
df_final
```

```python
# save the dataframe
df_final = df_final.compute()

save_path = "../data/interim/processing_data.csv"
df_final.to_csv(save_path, index=False)
```

```python
# train the standard scaler
scaler = StandardScaler()
df_reader = pd.read_csv(save_path, chunksize=100000, usecols=["pickup_latitude", "pickup_longitude"])

for chunk in df_reader:
    # fit the scaler
    scaler.partial_fit(chunk)

scaler
```

```python
# train the miniBatchKMeans model
df_reader = pd.read_csv(save_path, chunksize=100000, usecols=["pickup_latitude", "pickup_longitude"])
mini_batch = MiniBatchKMeans(n_clusters=30, n_init=10, random_state=42)

for chunk in df_reader:
    # scale the chunk
    scaled_chunk = scaler.transform(chunk)
    # train the model
    mini_batch.partial_fit(scaled_chunk)

mini_batch
```

```python
# centroids of the model

mini_batch.cluster_centers_
```

```python
scaler.inverse_transform(mini_batch.cluster_centers_)
```

```python
# perform predictions and assign clusters

location_subset = df_final[df_final.columns[1:]]
location_subset
```

```python
# scale the input data

scaled_location_subset = scaler.transform(location_subset)
scaled_location_subset
```

```python
# get the cluster predictions

cluster_predictions = mini_batch.predict(scaled_location_subset)
cluster_predictions
```

```python
# save the cluster/region predictions in data

df_final['region'] = cluster_predictions
df_final
```

```python
# drop the latitude and logitude columns from data as we already have region column and location information is not needed

time_series_data = df_final.drop(columns=["pickup_latitude", "pickup_longitude"])
time_series_data
```

```python
# save the time series and region dataset

save_path = "../data/interim/time_series.csv"
time_series_data.to_csv(save_path, index=False)
time_series_data.dtypes
```

```python
# set the time series column as the index of the dataframe

time_series_data.set_index('tpep_pickup_datetime', inplace=True)
time_series_data
```

```python
# group the time series data based on regions to calculate the demand for each region and each 15 minutes interval

region_grp = time_series_data.groupby("region")
region_grp
```

```python
# check for missing values
# No missing values is found
time_series_data.isna().sum()
```

```python
# resample the time series in 15 minute intervals

resampled_data = (
    region_grp['region']
    .resample("15min")
    .count()
)
resampled_data
```

Now, we have number of pickups in 15 minute time series data for each region.

```python
# setting the column name and resetting from multi-index series to single index dataframe
resampled_data.name = "total_pickups"
resampled_data = resampled_data.reset_index(level=0)
resampled_data
```

```python
# we have 3668 zeros in the data
(resampled_data['total_pickups'] == 0).sum()
```

```python
# replacing the zeros in the data with a small value (epsilon) becuase 0 in demominator during MAPE calculation 
# will lead to very big values
epsilon_val = 10
resampled_data.replace({'total_pickups': {0 : epsilon_val}}, inplace=True)
(resampled_data['total_pickups'] == 0).sum()
```

# Smoothing

## Moving Average

```python
resampled_data['total_pickups'].rolling(window=4).mean()
```

```python
window_values = list(range(3, 11, 1))
window_values
# Here, we will be using a rolling window of 3 to 10
```

```python
def calculate_best_window_value(windows):
    for window in windows:
        ind = window - 1
        # here, we are using ind because the rolling mean will have NaN values for the first (window-1) entries,
        # so we need to skip those when calculating the error.
        y_pred = resampled_data['total_pickups'].rolling(window=window).mean().values[ind:]
        y = resampled_data['total_pickups'].values[ind:]
        error = mean_absolute_percentage_error(y, y_pred)
        print(f"For window value {window}, the MAPE is {error:.2f}")
```

```python
calculate_best_window_value(window_values)
```

We can notice that MAPE is low when window size is low thus low window size will make recent data points more important.

## EWMA

```python
resampled_data['total_pickups'].ewm(alpha=0.9).mean()
# NOTE: EWMA don't give Nan values thus no use of ind variable here
```

```python
smoothing_values = np.arange(0.2, 1, 0.1)
smoothing_values
```

```python
def calculate_best_smoothing_value(values):
    y = resampled_data['total_pickups'].values
    for value in values:
        y_pred = resampled_data['total_pickups'].ewm(alpha=value).mean()
        error = mean_absolute_percentage_error(y, y_pred)
        print(f"For smoothing value {value:.1f}, the MAPE is {error:.2f}")
```

```python
calculate_best_smoothing_value(smoothing_values)
```

We can notice that high alpha/smoothing factor is reducing the MAPE thus giving more weightage to the recent data points.

```python
# dataset with pickup smoothing applied
# for first value, prediction will be same as actual value thus no need of shift here
# there after prediction will be more close to recent data points then farther data points
resampled_data["avg_pickups"] = resampled_data['total_pickups'].ewm(alpha=0.4).mean().round()
resampled_data
```

```python
# dataset with pickup smoothing applied (shifted by 1 to avoid data leakage)

resampled_data["avg_pickups"] = resampled_data['total_pickups'].ewm(alpha=0.4).mean().round()
resampled_data
```
