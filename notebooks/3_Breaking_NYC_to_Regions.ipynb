Dask does not have a dedicated `MiniBatchKMeans` class built natively into its machine learning library (`dask-ml`). Thus, we will use `Pandas` with `chunk_size` to load the data in chunks and train using `partial_fit` method of `MiniBatchKMeans` from `sklearn` library to perform the clustering on chunks of our dataset.
Then distance between every centroids will be calculated and sorted to get the nearest 8 regions for each region and assign the neighboring regions.

We will loop for different k values such that distance between centroids is around 1 mile.
Create a grid of centroids values and calculate the distance between each centroids and sort the distances for each regions to bring in ascending order. We will only consider out of 8 regions.
Average of these distances between 8 regions/neighbors and the region will be around 1 mile, then flag as good region, else flag as bad region. We will choose the k value which has more good regions and less bad regions.

```python
# train the standard scaler on each chunk of the whole dataset

scaler = StandardScaler()

df_reader = pd.read_csv(data_path, chunksize=100000)

for chunk in df_reader:
    # fit the scaler
    scaler.partial_fit(chunk)

scaler
```

```python
# now, we got the 2 scaler std and 2 mean

print("The mean values are ", scaler.mean_)
print("The sigma values are ", scaler.scale_)
```

```python
# for k = 30
df_reader = pd.read_csv(data_path, chunksize=100000)
mini_batch = MiniBatchKMeans(n_clusters=30, n_init=10, random_state=42, batch_size=1024)

for chunk in df_reader:
    # scale the chunk with trained/fitted scaler
    scaled_chunk = scaler.transform(chunk)
    # partial fitting/training the model
    mini_batch.partial_fit(scaled_chunk)
    
# scaled centroid values
centroids = mini_batch.cluster_centers_
centroids
```

```python
# inverse transform the centroids to get the original values
centroids = scaler.inverse_transform(mini_batch.cluster_centers_)
centroids.shape
# 30 regions with 2 values (latitude and longitude) for each region
```

```python
centroids
```

```python
# We are using haversine_distance because KMeans calculate the Euclidean distance between the points.
# But, we need to calculate the distance between the points on the earth surface.
# So, we need to use haversine_distance to calculate the distance between the points on the earth surface. 

def haversine_distance(lat1:float, lon1:float, lat2:float, lon2:float):
    """
    Calculate haversine distances between two points given their latitude and
    longitude coordinates and return the distance in miles.
    """

    lon1, lat1, lon2, lat2 = map(np.radians, [lon1, lat1, lon2, lat2])

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = np.sin(dlat / 2.0)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2.0)**2

    c = 2 * np.arcsin(np.sqrt(a))

    earth_radius = 3958.8
    distance = earth_radius * c
    return distance
```

```python
# calculate matrix and distances
all_distances = [] # row
for i in centroids:
    distances = [] # column
    for j in centroids:
        distance = haversine_distance(i[0], i[1], j[0], j[1])
        distances.append(distance)
    all_distances.append(distances)
all_distances = np.array(all_distances)
```

```python
def select_optimum_k(k_values):
    """
        Function to select the optimum k value for clustering based on the average distance between the centroids of the clusters.
    """
    for k in k_values:
        df_reader = pd.read_csv(data_path, chunksize=100000)
        mini_batch = MiniBatchKMeans(n_clusters=k, n_init=10, random_state=42)
        
        for chunk in df_reader:
            # scale the chunk
            scaled_chunk = scaler.transform(chunk)
            # train the model
            mini_batch.partial_fit(scaled_chunk)
            
        # centroid values
        centroids = scaler.inverse_transform(mini_batch.cluster_centers_)
        # calculate distances
        all_distances = []
        for i in centroids:
            distances = []
            for j in centroids:
                distance = haversine_distance(i[0], i[1], j[0], j[1])
                distances.append(distance)
            all_distances.append(distances)
        all_distances = np.array(all_distances)
        # sort array
        sorted_distances = np.sort(all_distances, axis=-1)
        # nearby regions
        region_count = 8
        # selected distances
        selected_distances = sorted_distances[:, 1:region_count+1]
        # average distances
        avg_distances = selected_distances.mean(axis=1)
        # count averages that fit criteria of being between 1.0 and 1.5 miles
        counter = ((avg_distances >= 1.0) & (avg_distances <= 1.5)).sum()
        percentage = ((avg_distances >= 1.0) & (avg_distances <= 1.5)).mean()
        print(f"For {k} clusters, the number of nearby regions that fits criteria are {counter} and is {percentage * 100:.1f}% regions of total")
```

```python
# select the optimum value of K
possible_k_values = np.arange(10, 100, 10)
select_optimum_k(k_values=possible_k_values)
```

We can notice that for 30 clusters, 9 clusters (30% of clusters) have average distance to nearby regions between 1.0 and 1.5 miles.
So, we are going to use k = 30 for our clustering.

# Clustering

```python
# train the model
k = 30
mini_batch = MiniBatchKMeans(n_clusters=k, n_init=10, random_state=42)
df_reader = pd.read_csv(data_path, chunksize=100000)

for chunk in df_reader:
    # scale the chunk
    scaled_chunk = scaler.transform(chunk)
    # train the model
    mini_batch.partial_fit(scaled_chunk)
```
